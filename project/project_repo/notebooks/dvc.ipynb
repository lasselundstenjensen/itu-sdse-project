```python
import pandas as pd

# Load directly from the raw GitHub URL
url = "https://raw.githubusercontent.com/Jeppe-T-K/itu-sdse-project-data/refs/heads/main/raw_data.csv"
df = pd.read_csv(url)

# 1. Convert date columns to datetime objects
date_cols = ['date_part', 'first_booking', 'last_seen']
for col in date_cols:
    df[col] = pd.to_datetime(df[col])

# 2. Handle missing target values (lead_indicator)
# If you are building a classifier, you usually drop rows with no target label
df_clean = df.dropna(subset=['lead_indicator'])

# 3. Check the result
print(df_clean.info())
print(df_clean['lead_indicator'].value_counts())
```
